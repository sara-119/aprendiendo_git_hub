## Contexto
Considere el siguiente `DataFrame` `df`, con información de 12 empleados de una empresa. Ejecute la siguiente celda para crearlo.

```python
import pandas as pd
import numpy as np

datos = {
    "Nombre": ["Mariana", "Esteban", "Camilo", "Juliana", "Nicolas", "Paula",
               "Sebastian", "Laura", "Andres", "Catalina", "Esteban", "Manuela"],
    "Edad": [24, 29, 35, 22, 31, 27, 40, 26, 33, 29, 29, 24],
    "Genero": ["Mujer", "H", "Hombre", "M", "Hombre", "Mujer",
               "H", "Mujer", "Hombre", "M", "H", "Mujer"],
    "Ciudad": ["Bogota", "Medellin", "Cali", "Bogota", "Bucaramanga", "Medellin",
               "Bogota", "Cali", "Pasto", "Bogota", "Medellin", "Bogota"],
    "Departamento": ["Marketing", "Ventas", "IT", np.nan, "Ventas", "Marketing",
                     "IT", np.nan, "Finanzas", "Marketing", "Ventas", "Finanzas"],
    "Salario": [3.8, 4.2, 6.0, 3.2, 4.0, 3.9, 6.3, 3.7, 4.5, 55.0, 4.2, 3.6],
    "AñosExperiencia": [2, 5, 10, -3, 7, 4, 15, 3, 9, 6, 5, 2],
}

df = pd.DataFrame(datos)
df
```

## Subsección 1: Limpieza de datos

**1.** La variable **Departamento** tiene valores faltantes (`NaN`) en las filas de **Juliana** y **Laura**. Escriba el código en pandas que impute estos valores utilizando la **moda** de la variable.

```python
df["Departamento"].mode
df["Departamento"] = df["Departamento"].fillna(df["Departamento"].mode)

df
```

**2.** La variable **AñosExperiencia** tiene un valor negativo en la fila de **Juliana** (`-3`), producto de un error de digitación. Escriba el código en pandas que corrija este valor, convirtiéndolo en positivo.

```python
# Escriba aquí su código
df.loc[df["Nombre"] == "Juliana", "AñosExperiencia"] = abs(df.loc[df["Nombre"] == "Juliana", "AñosExperiencia"])
df
```

**3.** La variable **Genero** presenta inconsistencias: aparece registrada como `"Hombre"`, `"H"`, `"Mujer"` y `"M"`. Escriba el código en pandas que estandarice esta variable para que solo queden los valores `"Hombre"` y `"Mujer"`.

```python
# Escriba aquí su código
df["Genero"] = df["Genero"].replace({"H": "Hombre","M": "Mujer"})

df
```

**4.** El `DataFrame` contiene una fila **duplicada**: el empleado **Esteban** aparece registrado dos veces de forma idéntica. Escriba el código en pandas que detecte y elimine las filas duplicadas.

```python
# Escriba aquí su código
df = df.drop_duplicates()
df
```

**5.** La variable **Salario** de **Catalina** tiene un valor atípico (`55.0`), muy alejado del resto de los datos. Corríjalo asumiendo que se trata de un error de digitación (un cero de más).

```python
# Escriba aquí su código
df.loc[df["Nombre"] == "Catalina", "Salario"] = ( df.loc[df["Nombre"] == "Catalina", "Salario"] / 10)
df
```

**6.** El registro de **Manuela** está desactualizado: en realidad ella tiene **8 años** de experiencia. Escriba el código en pandas que corrija ese dato.

```python
# Escriba aquí su código
df.loc[df["Nombre"] == "Manuela", "AñosExperiencia"] = 8
df
```
